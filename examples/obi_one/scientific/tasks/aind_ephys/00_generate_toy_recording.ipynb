# 0. Generate a Toy NWB Recording

Writes a 10 s, 32-channel ground-truth recording of one probe as an NWB file, the input the dispatch stage (notebook 01) reads. The probe geometry goes into the NWB electrodes table, which is where the pipeline reads it back from. The ground-truth sorting is saved beside it so notebook 03 can score its sorting against it. Everything goes under `obi-output/`, outside the repo.

## Write the NWB file and the ground truth

```python
from pathlib import Path

from obi_one.scientific.tasks.aind_ephys.toy_recording import write_toy_nwb

obi_output = (Path.cwd() / "../../../../../../obi-output").resolve()
toy_folder = obi_output / "00_toy_example_recording"
nwb_path = toy_folder / "toy_recording.nwb"

ground_truth = write_toy_nwb(nwb_path, duration_s=10.0, num_channels=32, num_units=10)
ground_truth.save(folder=toy_folder / "ground_truth_sorting", overwrite=True)
print("NWB file:", nwb_path)
ground_truth
```

## Read the recording back as the pipeline will

```python
import spikeinterface.extractors as se

recording = se.read_nwb_recording(nwb_path, electrical_series_path="acquisition/ElectricalSeriesRaw")
print(recording)
print("probe:", recording.get_probe())
```

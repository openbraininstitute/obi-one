# Basic connectivity plots

Example notebook on how to generate basic connectivity plots from a ConnectivityMatrix object

```python
from obi_one.core.deserialize import deserialize_obi_object_from_json_file
from obi_one.core.path import NamedPath
from obi_one.core.run_tasks import run_tasks_for_generated_scan
from obi_one.core.scan_generation import GridScanGenerationTask

from obi_one.scientific.tasks.basic_connectivity_plots.config import BasicConnectivityPlotsScanConfig
from pathlib import Path
```

```python
matrix_path_prefix = Path("../../../../data/connectivity_matrices")
```

```python
basic_connectivity_plots_scan_config = BasicConnectivityPlotsScanConfig(
                    initialize=BasicConnectivityPlotsScanConfig.Initialize(
                        matrix_path=[NamedPath(name="N_10__top_nodes_dim6", path=str(matrix_path_prefix / "N_10__top_nodes_dim6" / "connectivity_matrix.h5")),
                                     NamedPath(name="N_10__top_rc_nodes_dim2_rc", path=str(matrix_path_prefix / "N_10__top_rc_nodes_dim2_rc" / "connectivity_matrix.h5"))],
                        plot_formats=("png",),  # sub-tuple of ('png', 'pdf', 'svg'), if not specified all are plotted
                        rendering_cmap="custom",  # "custom" + .csv file OR Python cmap, e.g., "tab10"
                        rendering_color_file="colors_tab10.csv",
                        # plot_types=("nodes",), # sub-tuple of ('nodes', 'connectivity_global', 'connectivity_pathway'), if not specified all are plotted
                        # dpi=int default 300
                    )
)

grid_scan = GridScanGenerationTask(form=basic_connectivity_plots_scan_config, output_root="../../../../../../obi-output/basic_connectivity_plots/grid_scan", coordinate_directory_option="VALUE")
grid_scan.execute()
```

```python
run_tasks_for_generated_scan(grid_scan)
```

```python
# Display the intermediary data
grid_scan.multiple_value_parameters(display=True)
grid_scan.coordinate_parameters(display=True)
```
